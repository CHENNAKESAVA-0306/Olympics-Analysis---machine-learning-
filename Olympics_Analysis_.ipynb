# Model Fitting

```python
df = df.reset_index(drop=True)
```

```python
df
```

```python
# train and test split data
from sklearn.model_selection import train_test_split
x_train,y_train,x_test,y_test = train_test_split(x,y,test_size = 0.25,random_state = 42)  # test data  = 0.25(25%) and train data (75%)
print(x_train.shape)
print(x_test.shape)
print(y_train.shape)
print(y_test.shape)
```

```python
x_train
```

```python
x_test
```

```python
# Independent features and dependent features
x = df[['Year','Gender','Event']]  # "x" represents the Independent features or variables
y = df['Medal']   # "y" represents the dependent features or variables


```

```python
# Convert 'Medal' column to numerical using LabelEncoder
label_encoder = LabelEncoder()
y = label_encoder.fit_transform(y)
```

```python
y
```

```python

from sklearn.preprocessing import StandardScaler
scaler = StandardScaler()
x_train = scaler.fit_transform(x_train)
x_test = scaler.transform(x_test)
```

```python
x_train
```

```python
x_test
```

```python
from sklearn.linear_model import LinearRegression
# Linear Regression Model
print(" LinearRegression Model")
lr = LinearRegression(n_jobs=-1)
lr.fit(x_train, y_train)  # Now y_train contains numerical labels
```

```python
print('Intercept',lr.intercept_)
```

```python
# Convert 'Medal' column to numerical using pd.factorize before plotting

df['Medal_Numeric'] = pd.factorize(df['Medal'])[0]
```

```python
# regression plot between the "Gender"and "Medal"
sns.regplot(x=df['Gender'],y=df['Medal_Numeric'])
  # in the below graph shows the 0.0 represents the "Men" and 1 represnts the "women"
```

```python
sns.regplot(x=df['Year'],y=df['Medal_Numeric']) # regression plot between the "year" and "Medals "
```

```python
## cross validation and its purpose to reduce the overfitting
from sklearn.model_selection import cross_val_score
validation_score = cross_val_score(lr,x_train,y_train,scoring = 'neg_mean_squared_error',cv=2)
```

```python
validation_score
```

```python
np.mean(validation_score)
```

```python
# predication
y_pred =lr.predict(x_test)
```

```python
y_pred
```

# Performance Matrices

```python
## Performance Metrics

from sklearn.metrics import mean_squared_error,mean_absolute_error
mse = mean_squared_error(y_test,y_pred)
mae = mean_absolute_error(y_test,y_pred)
print(mse)
print(mae)
rmse = np.sqrt(mse)      # root mean squared error
print(rmse)
```

```python
from sklearn.metrics import r2_score
accuracy_score = r2_score(y_test,y_pred)
print(accuracy_score)
```

```python
# adjusted r2 error

adjusted_r2 = 1-(1-r2_score(y_test,y_pred))*(len(y_test)-1)/(len(y_test)-x_test.shape[1]-1)   # formula
print(adjusted_r2)
```

```python
# residuals or error

residuals = y_test-y_pred
residuals                              # error can be defined as the difference between y_test and y_predication
```

```python
## Plot this residuals
sns.displot(residuals,kind = 'kde')
```

```python
## scatter plot with respect to prediction and residuals

plt.scatter(y_pred,residuals)
```

```python
#correlation matrix using numeric columns

numeric_cols = df.select_dtypes(include=[np.number])
correlation_matrix = numeric_cols.corr()
plt.figure(figsize=(10,6))
sns.heatmap(correlation_matrix,annot = True,cmap='coolwarm',linewidth =0.5)
plt.title('correlation matrix')
plt.show()
```
